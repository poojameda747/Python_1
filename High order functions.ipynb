8.  Given a list of integers, use map() with id() to print the memory address
of each element.
Example: [10, 350, 10, 350, 20] — explain why some addresses repeat.

```python
l = [10,350,10,350,20]

result = list(map(lambda x: id(x), l))

print(result)
```

10.  Given a list of numbers:
[5, 10, 15, 20, 25, 30]
Perform the following in a single pipeline:
• Use map() to square each number
• Use filter() to keep only numbers divisible by 5
• Use reduce() to calculate the sum of remaining number

```python
l = [5,10,15,20,25,30]
result = list(map(lambda x:x*x,l))
print(result)
```

```python
l = [5,10,15,20,25,30]
result = list(filter(lambda x:x%5==0,l))
print(result)
```

```python
from functools import reduce
l = [5,10,15,20,25,30]
result = reduce(lambda x,y:x+y,l)
print(result)
```

```python
from functools import reduce
l = [5,10,15,20,25,30]
result= reduce(lambda x,y:x+y,
    filter(lambda x:x%5==0,
    map(lambda x:x*x,l)
           )
               )
print(result)
```
