```python
#Given two lists:
#a = [1, 2, 3, 4] b = [10, 20, 30, 40]
#Use map() with a lambda to create a new list #containing the sum of corresponding
#elements.
#What ha#ppens if the lists are of unequal length?

a = [1,2,3,4]
b = [10,20,30,40]
l = list(map(lambda x, y: x+y, a,b))
print(l)
```

```python
#if list are different lengths
a = [10,20,20.10]
b = [2,3]
l = list(map(lambda x,y:x+y, a,b))
print(l)
```

```python
nums = [12, 15, 7, 18, 20, 21, 25]

result = list(filter(lambda x: (x % 3 == 0 or x % 5 == 0) and not (x % 3 == 0 and x % 5 == 0), nums))

print(result)
```

```python
from functools import reduce
li = [1,2,3,4]
result = reduce(lambda x,y:x+y, li,10)
print(result)
```

```python
nums = [[1, 2], [3, 4], [5, 6]]
result = list(map(lambda x: x.append(10), nums))
print("Result:", result)
print("Nums:", nums)
```
